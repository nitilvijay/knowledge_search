```python
import os
os.environ["FLAGS_use_onednn"] = "0"

import cv2
import numpy as np
from paddleocr import PaddleOCR
from tqdm import tqdm

# Initialize OCR
ocr = PaddleOCR(
    lang="en",
    enable_mkldnn=False,
    use_doc_orientation_classify=False,
    use_doc_unwarping=False,
    use_textline_orientation=False,
)

image_path = "News/IMG-20260517-WA0006.jpg"
print(f"Loading image: {image_path}")
img = cv2.imread(image_path)
if img is None:
    raise FileNotFoundError(f"Could not load image at path: {image_path}")

# Run OCR prediction
result = ocr.predict(img)
```

```python
texts = result[0]["rec_texts"]
texts_string = " ".join(texts)

print(texts)
```

```python
from sentence_transformers import SentenceTransformer, util
import math

model = SentenceTransformer('BAAI/bge-small-en-v1.5')
len_texts = len(texts_string.split())
num_chunks = len_texts / 350   #context window of model is 384
chunks = []
embeddings = []

start_idx = 0
end_idx = 350
overlap_portion = 350*0.3

while start_idx < len_texts:
    
    #ignore the last chunk if it has less than 50 words
    if len_texts - start_idx < 50 and num_chunks > 1:
        break
    chunk = " ".join(texts_string.split()[start_idx:end_idx])
    chunks.append(chunk)
    
    start_idx = math.ceil(end_idx-overlap_portion)
    end_idx = math.ceil(start_idx+350)

for text in chunks:
    embeddings.append(model.encode(text))
    


# embedding1 = model.encode("Use of AI in warfare as a weapon")

# for i in embeddings:
#     similarity = util.cos_sim(embedding1, i)
#     print(f"Similarity score: {similarity.item():.4f}")

```
